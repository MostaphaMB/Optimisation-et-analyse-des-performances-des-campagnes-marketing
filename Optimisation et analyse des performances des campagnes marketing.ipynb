***Étape 2 :*** Tests Statistiques

- Impact des Campagnes sur les Ventes (T-test):

```python
from scipy import stats
df['MntTotal'] = df[['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntBakedProds']].sum(axis=1)

# H0 : Les moyennes sont identiques
# H1 : Ceux qui acceptent la campagne (Response=1) dépensent plus en moyenne.
group_yes = df[df['Response'] == 1]['MntTotal']
group_no = df[df['Response'] == 0]['MntTotal']
# Calcul de T-test et P_value
t_stat, p_val_t = stats.ttest_ind(group_yes, group_no)
print(f"T-Test (Response vs MntTotal) :")
print(f" - Moyenne acceptation : {group_yes.mean():.2f}")
print(f" - Moyenne refus : {group_no.mean():.2f}")
if p_val_t < 0.05:
    print(f" - P-value : {p_val_t:.4e} (Significatif si < 0.05)\n")
    print(f"Donc, ceux qui acceptent la campagne (Response=1) dépensent plus en moyenne.")
else:
    print(f" - P-value : {p_val_t:.4e} (On peut accepter l'hypothése H0 : Les moyennes sont identiques")

print(p_val_t)
group_no
# df
```

- Lien entre les Pays et l'acceptation de la campagne (Test Chi-2 ):

```python
# H0 : Les variable sans independentes (pas de relation)
# H1 : Il ya une relation entre les variables (Pays et Acceptation de la campagne)
print(f"Test Chi-2 (Country vs Response) :")
# Calcul de Test Chi-2 et P_value
contingency = pd.crosstab(df['Country'], df['Response'])
chi2, p_val_chi2, dof, expected = stats.chi2_contingency(contingency)

if p_val_chi2 < 0.05:
    print(f" - P-value : {p_val_chi2:.4f} (Lien significatif si < 0.05)\n")
    print(f"Donc, Il ya une relation entre les variables (Pays et Acceptation de la campagne).")
else:
    print(f""" - P-value : {p_val_chi2:.4f} (Lien non significatif si > 0.05) 
         On peut accepter l'hypothése H0 : Les variable sans independentes (pas de relation) 
          """)
```
