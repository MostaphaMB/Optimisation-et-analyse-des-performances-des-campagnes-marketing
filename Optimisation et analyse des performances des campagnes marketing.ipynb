Visualisation des données :

```python

df.hist(column=['MntWines',
       'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts',
       'MntBakedProds'],figsize=(13,8))

plt.show()
```

```python
# Liste des colonnes de dépenses
mnt_cols = ['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntBakedProds']

# Boxplot pour voir les outliers de dépenses
plt.figure(figsize=(13, 8))
sns.boxplot(data=df[mnt_cols])
plt.title("Distribution des dépenses par catégorie")
plt.show()
```

```python
acceptances = df[['AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4', 'AcceptedCmp5','Response']].sum()
plt.figure(figsize=(10,6))
plt.bar(acceptances.index, acceptances.values, color='skyblue')
plt.title('Nombre d\'acceptations par campagne')
plt.xlabel('Campagne')
plt.ylabel('Nombre d\'acceptations')
plt.show()
```

Interprétation: Nous remarquant que le nombre d'accéptation de la compagne publicitaire '2' et le plus bas ,et la compagne publicitaire '4' a plus d'accéptation.

```python
# Matrice de corrélation entre les variables numériques
plt.figure(figsize=(16, 10))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="seismic", linewidths=0.5, fmt=".2f")
plt.title("Matrice de Corrélation")
plt.show()
```

```python
# corrélation entre le revenu et montant d'achats de viande
plt.figure(figsize=(12, 6))
sns.scatterplot( data=df, x=' Income ', y='MntMeatProducts')
plt.title("Corrélation entre le revenu et les achats de viande")
plt.xlabel("Revenu")
plt.ylabel("Achats de viande")
plt.show()
```

Interprétation : Nous observons une forte corrélation entre le revenu et les achats de viande

```python
# corrélation entre le Nombre d'achats avec une réduction et le nombre de visites du site web
plt.figure(figsize=(12, 6))
sns.scatterplot( data=df, x='NumWebVisitsMonth', y='NumDealsPurchases')
plt.title("Corrélation entre le nombre d'achats avec une réduction et le nombre de visites du site web")
plt.xlabel("Nombre d'achats avec une réduction")
plt.ylabel("Nombre de visites du site")
plt.show()
```
